# write a program to identify whether the user entered number is prime number or not

```python
num = int(input("Enter a Number"))
divisor = num-1
flag=1
while divisor>=2:
    if num%divisor==0:
        flag=0
        break
    else:
        pass
    divisor=divisor-1
if flag==1:
    print("num is prime")
else:
    print("num is not prime")
    
```

# NESTED LOOP

```python
i=1
while i<=3:
    j=1
    while j<=2:
        print(i,j)
        j=j+1
    i=i+1
       
```

# WAP to find maximum in a list ,without using max function

```python
list1 = [45,34,23,60,11]
max_num=0


for i in list1:
    if i>max_num:
        max_num=i
print("Maximum in the list is: ",max_num)
```
