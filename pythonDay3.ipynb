# write a program to identify whether the user entered number is prime number or not

```python
num = int(input("Enter a Number"))
divisor = num-1
flag=1
while divisor>=2:
    if num%divisor==0:
        flag=0
        break
    else:
        pass
    divisor=divisor-1
if flag==1:
    print("num is prime")
else:
    print("num is not prime")
    
```

# NESTED LOOP

```python
i=1
while i<=3:
    j=1
    while j<=2:
        print(i,j)
        j=j+1
    i=i+1
       
```

```python
i=1
while i<=3:
    j=1
    while j<=2:
        k=1
        while k<=2:
            print(i,j,k)
            k=k+1
        j=j+1
    i=i+1
       
```

```python
i=1
while i<=3:
    j=1
    while j<=2:
        if i==2:
            break
        print(i,j)
        j=j+1
    i=i+1
```

# CONTINUE

```python
i=1
while i<=10:
    if i==4:
        i=i+1
        continue
    print(i)
    i=i+1
```

```python
i=1
while i<=10:
    if i==4:
        i=i+1
        continue
    print(i,end=" ")
    i=i+1
```

```python
text = "abcd"
i=0
while i<=3:
    if i==2:
        i=i+1
        continue
    print(text[i])
    i=i+1
```

```python
i=1
while i<=50:
    if i==i*7:
        i=i+1
        continue
    print(i,end=" ")
    i=i+1
```
